W zdaniu 6 przycięcie "mięsiącach" do "miesiąc" może powodować mylne wrażenie o rzeczywistej ilości mięsięcy (kilka vs. jeden). W zdaniu 9 "jeśl" nie jest formą bazową ani słowem.

Stemming redukuje rozmiar słownika o 32.47% w stosunku do surowego tekstu.

### Zadanie 3



Najlepszy wynik osiągnęła metoda Stemming, co może być spowodowane tworzeniem przez tą metodę bardziej zwięzłej reprezentacji dla krótkich tekstów, jak recenzje, niż lematyzacja. Różnica między stemmingiem a lemantyzacją jest istotna. Stemming często drastycznie ucina końcówki słów, prowadząc do powstania nieistniejących form, co może powodować utratę informacji semantycznej. Lematyzacja redukuje słowa do ich słownikowej formy podstawowej, co pozwala w sporej ilości przypadków zachować poprawne znaczenie słów i lepiej łączyć różne formy gramatyczne tego samego wyrazu w jeden wektor cech.
